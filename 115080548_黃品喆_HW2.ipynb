## **Basis of Numpy Library**

Numpy is the core library for scientific computing in Python. It provides a high-performance multidimensional array object and tools for working with these **arrays**.

To use Numpy, we first need to import the `numpy` package:

```python
# import library
import numpy as np
```

### **Arrays**

- A numpy array is a grid of values, all of the same type, and is indexed by a tuple of nonnegative integers.

- The number of dimensions is the rank of the array;

- The shape of an array is a tuple of integers giving the size of the array along each dimension.

```python
# create 1d array
array_1d = np.array([0, 1, 2, 3, 4, 5])
print(array_1d)
```

```python
# print shape
print(array_1d.shape)
```

```python
# create 2d array
array_2d = np.array([
    [1, 2, 3],
    [4, 5, 6]
])
print(array_2d)
```

```python
# print shape
print(array_2d.shape)
```

```python
# create 3d array
array_3d = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])
print(array_3d)
```

```python
# print shape
print(array_3d.shape)
```

### **Array Indexing**

```python
# print the second element in your 1d array
print(array_1d[1])
```

```python
# print the first row second element in your 2d array
print(array_2d[0, 1])
```

```python
# print the first rank first row second element in your 3d array
print(array_3d[0, 0, 1])
```

### **Array Operations**

```python
# create 2 2d-array
x = np.array([[1, 2], [3, 4]], dtype=np.float64)

y = np.array([[5, 6], [7, 8]], dtype=np.float64)

# Elementwise sum; both produce the same array
print(x + y)
print(np.add(x, y))
```

```python
# Elementwise difference; both produce an array
print(x - y)
print(np.subtract(x, y))
```

```python
# Elementwise division; both produce an array
print(x / y)
print(np.divide(x, y))
```

```python
# Elementwise product; both produce the array
print(x * y)
print(np.multiply(x, y))
```

```python
# Dot product or matrix multiplication
print(x.dot(y))
print(np.dot(x, y))
print(x @ y)
```

```python
# read file
import cv2
from google.colab.patches import cv2_imshow
import matplotlib.pyplot as plt

filename = '/content/drive/MyDrive/DL HW/115080548_黃品喆_HW2.png' # your file path

img = cv2.imread(str(filename), cv2.IMREAD_GRAYSCALE)

print("Image shape:", img.shape)
print("Data type:", img.dtype)

plt.figure(figsize=(5, 5))
plt.imshow(img, cmap="gray", vmin=0, vmax=255)
plt.title("Original noisy image")
plt.axis("off")
plt.show()
```

```python
# Rotate
rotated = cv2.rotate(img, cv2.ROTATE_180)

plt.figure(figsize=(5, 5))
plt.imshow(rotated, cmap="gray", vmin=0, vmax=255)
plt.title("After 180° rotation")
plt.axis("off")
plt.show()
```

```python
# Denoise by filter
denoised = cv2.medianBlur(rotated, 5)

plt.figure(figsize=(5, 5))
plt.imshow(denoised, cmap="gray", vmin=0, vmax=255)
plt.title("Denoised image (5x5 median filter)")
plt.axis("off")
plt.show()
```
